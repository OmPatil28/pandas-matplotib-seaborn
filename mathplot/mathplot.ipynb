```python
import matplotlib.pyplot as plt
```

```python
x = [1,2,3,4,5]
y = [10,5,7,11,13]

plt.plot(x,y)
plt.show()
```

```python
# customize chart

plt.plot(x, y, color='pink', marker='o', linestyle='dashed', linewidth=2, markersize=12)

plt.title("show the info about chart")
plt.xlabel("info about x-axis")
plt.ylabel("iinfo about y-axis")
plt.show()
```

```python
x = [1,2,3,4,5]
y1 = [10,5,15,20,30]
y2 = [20,15,10,25,35]

plt.plot(x, y1, label='sales 2024')
plt.plot(x, y2, label='sales 2025')
plt.title('yoy sales')
plt.xlabel("month")
plt.ylabel("sale")
plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt

#data set 1
categories = ['Mon','Tue','Web','Thu','Fri']
sales = [10,5,15,20,30]

#data set 2
y1 = [10,20,30,40,50]
y2 = [20,30,35,45,55]



#first plot -bar chart
plt.subplot(2,1,1)
plt.bar(categories,sales)
plt.title('daily sales')
plt.xlabel('week sales')
plt.ylabel('sales')

#secound plot

plt.subplot(2,1,2)
plt.scatter(y1,y2)
plt.title('user exapmle')
plt.xlabel('user1')
plt.ylabel('user2')

plt.show()
```

```python
import pandas as pd

data = {
    'month' : ['JAN', 'FEB','MARCH', 'APRIL'],
    'SALES': [12000,23000, 34000, 55000]
}
df = pd.DataFrame(data)
df
```

```python
plt.bar(df['month'], df['SALES'])
plt.title('mathplot with pandas')
plt.xlabel('month')
plt.ylabel('sales')

plt.show()
```
